```python
import polars as pl
from coppertop.pipe import *
diffRows = None
from coppertop.dm.core import diffRows, collect
from coppertop.dm.pp import PP
```

```python
'sig1:' >> PP
sig1 = diffRows >> sig >> collect >> PP

from coppertop.dm.polarframe import diffRows

'\nsig2:' >> PP
sig2 = diffRows >> sig >> collect >> PP

@coppertop
def diffRows(f: int) -> pl.DataFrame:
    numericColNames = [cn for cn, t in f.schema.items() if t.is_numeric()]
    return f.with_columns(pl.col(numericColNames) - pl.col(numericColNames).shift(1))[1:]

'\nsig3:' >> PP
sig3 = diffRows >> sig >> collect >> PP

assert len(sig2) > len(sig1)
```
